Esse código gera dados fictícios para inserir na tabela `ingredientes` do banco, usando Python e a biblioteca Random para simular registros realistas.

## Imports

```python
# Importações necessárias
import random as rd
```

## Guardando a semente
Guardar a semente faz com que os dados gerados sejam sempre os mesmos

```python
rd.seed(0)
```

## Funções

### Função pluralizar

Pluraliza unidades de medida que possuem forma plural, considerando a quantidade.

```python
def pluralizar(unidade, quantidade):
    plurais = {
        "colher": "colheres",
        "colher de sopa": "colheres de sopa",
        "xícara": "xícaras",
        "pitada": "pitadas",
        "dente": "dentes",
        "barra": "barras",
        "fatia": "fatias",
        "unidade": "unidades"
    }
    if quantidade > 1 and unidade in plurais:
        return plurais[unidade]
    else:
        return unidade
```

### Função para gerar as quantidades dos ingredientes

```python
def gerar_quantidade(unidade):
    if unidade in ["g", "ml"]:
        quantidade = rd.randint(1, 500)
    elif unidade in ["kg", "l"]:
        quantidade = rd.randint(1, 2)
    elif unidade in ["colher", "colher de sopa", "xícara", "pitada", "unidade", "dente", "barra", "fatia"]:
        quantidade = rd.randint(1, 5)
    else:
        quantidade = 1

    unidade_formatada = pluralizar(unidade, quantidade)
    return f"{quantidade} {unidade_formatada}"
```

## Gerando 20 registros com dados falsos e imprimindo no formato SQL

```python
# Dicionário com ingredientes e possíveis unidades de medida
ingredientes_unidades = {
    "açúcar": ["g", "kg", "colher", "xícara"],
    "sal": ["g", "kg", "colher", "pitada"],
    "farinha de trigo": ["g", "kg", "xícara", "colher"],
    "óleo de soja": ["ml", "l", "colher de sopa"],
    "leite integral": ["ml", "l", "xícara"],
    "manteiga": ["g", "kg", "colher"],
    "ovos": ["unidade"],
    "amido de milho": ["g", "kg", "colher"],
    "canela": ["g", "colher", "pitada"],
    "chocolate meio amargo": ["g", "barra", "colher"],
    "aveia": ["g", "xícara", "colher"],
    "mel": ["g", "colher", "xícara"],
    "vinagre": ["ml", "colher", "l"],
    "pimenta-do-reino": ["g", "colher", "pitada"],
    "alho": ["g", "unidade", "dente"],
    "cebola": ["g", "unidade", "fatia"],
    "arroz": ["g", "kg", "xícara"],
    "feijão": ["g", "kg", "xícara"],
    "batata": ["g", "unidade", "fatia"],
    "queijo mussarela": ["g", "fatia", "unidade"]
}

nomes = list(ingredientes_unidades.keys()) # acessando as chaves que são os nomes dos ingredientes

# Começa a imprimir o comando SQL de inserção, 
# com os nomes das colunas da tabela 'ingredientes'

print("INSERT INTO ingredientes (nome, quantidade, id_nutrientes)\nVALUES",end="")

for i in range(20):
    nome = nomes[i]
    unidade_medida = rd.choice(ingredientes_unidades[nome])
    quantidade = gerar_quantidade(unidade_medida)
    id_nutrientes = i + 1

    # Imprime os valores formatados para inserir no banco
    if (i < 19):
        print(f"('{nome}', '{quantidade}', {id_nutrientes}),")
    else:
        print(f"('{nome}', '{quantidade}', {id_nutrientes});")
```

# Notebook para o gerador de script DataLoad da tabela nutrientes

Esse código gera dados fictícios para inserir na tabela `nutrientes` do banco, usando Python e a biblioteca Random para simular registros realistas.

---

## Imports

```python
# Importações necessárias
import random as rd
```

## Guardando a semente
Guardar a semente faz com que os dados gerados sejam sempre os mesmos

```python
rd.seed(0)
```

## Gerando registros aleatórios e imprimindo no formato SQL

```python
# Começa a imprimir o comando SQL de inserção, 
# com os nomes das colunas da tabela 'nutrientes'
print("INSERT INTO tabela_nutrientes (id_ingrediente,valor_energetico_kcal, carboidratos_g, acucares_totais_g, acucares_adicionados_g, proteinas_g, gorduras_totais_g, gorduras_saturadas_g, gorduras_trans_g, fibra_alimentar_g, sodio_mg, colesterol_mg, vitamina_a_mcg, vitamina_c_mg, vitamina_d_mcg, calcio_mg, ferro_mg, potassio_mg) \nVALUES ",end="")

for i in range(20):
    id_ingrediente = i + 1
    carboidratos_g = f"{rd.randint(0, 100)}g"
    acucares_totais_g = f"{rd.randint(0, 80)}g"
    acucares_adicionados_g = f"{rd.randint(0, 60)}g"
    proteinas_g = f"{rd.randint(0, 60)}g"
    gorduras_totais_g = f"{rd.randint(0, 50)}g"
    gorduras_saturadas_g = f"{rd.randint(0, 20)}g"
    gorduras_trans_g = f"{rd.randint(0, 2)}g"
    fibra_alimentar_g = f"{rd.randint(0, 15)}g"
    sodio_mg = f"{rd.randint(0, 1200)}mg"
    colesterol_mg = f"{rd.randint(0, 200)}mg"
    vitamina_A_mcg = f"{rd.randint(0, 900)}µg RE"
    vitamina_C_mg = f"{rd.randint(0, 100)}mg"
    vitamina_D_mcg = f"{rd.randint(0, 35)}µg"
    calcio_mg = f"{rd.randint(0, 500)}mg"
    ferro_mg = f"{rd.randint(0, 20)}mg"
    potassio_mg = f"{rd.randint(0, 1000)}mg"
    
# Imprime os valores formatados para inserir no banco
    if (i < 19):
            print(f"({id_ingrediente}, '{carboidratos_g}', '{acucares_totais_g}', '{acucares_adicionados_g}', '{proteinas_g}', '{gorduras_totais_g}', '{gorduras_saturadas_g}', '{gorduras_trans_g}', '{fibra_alimentar_g}', '{sodio_mg}', '{colesterol_mg}', '{vitamina_A_mcg}', '{vitamina_C_mg}', '{vitamina_D_mcg}', '{calcio_mg}', '{ferro_mg}', '{potassio_mg}') ,")
    else:
            print(f"({id_ingrediente}, '{carboidratos_g}', '{acucares_totais_g}', '{acucares_adicionados_g}', '{proteinas_g}', '{gorduras_totais_g}', '{gorduras_saturadas_g}', '{gorduras_trans_g}', '{fibra_alimentar_g}', '{sodio_mg}', '{colesterol_mg}', '{vitamina_A_mcg}', '{vitamina_C_mg}', '{vitamina_D_mcg}', '{calcio_mg}', '{ferro_mg}', '{potassio_mg}') ;")
```

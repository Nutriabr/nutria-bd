# Notebook para o gerador de script DataLoad da tabela da Tabela Nutricional

## Imports

```python
# Importações necessárias
import random as rd
```

## Guardando a semente
Guardar a semente faz com que os dados gerados sejam sempre os mesmos

```python
rd.seed(0)
```

## Gerando registros aleatórios e imprimindo no formato SQL

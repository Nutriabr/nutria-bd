## Funções

### Funcão para gerar porções

```python
def gerar_porcao(nome):
    solidos = [
        "Bolo de Açúcar e Canela",
        "Arroz com Feijão",
        "Purê de Batata",
        "Salada de Cebola e Queijo Mussarela",
        "Ovos Mexidos com Manteiga",
        "Pão de Mel com Chocolate Meio Amargo",
        "Panqueca de Farinha de Trigo e Leite Integral",
        "Cookie de Aveia e Mel",
        "Frango ao Vinagre com Pimenta-do-Reino",
        "Risoto de Arroz com Queijo Mussarela",
        "Feijoada Tradicional",
        "Biscoito de Amido de Milho",
        "Torta de Batata com Queijo Mussarela",
        "Bolo de Chocolate Meio Amargo",
        "Omelete de Ovo com Queijo Mussarela",
        "Pão de Queijo",
        "Batata Assada",
        "Panetone com Açúcar e Canela"
    ]

    liquidos = [
        "Molho de Tomate com Sal e Alho",
        "Caldo Verde"
    ]

    if (nome in solidos):
        return f"{rd.randint(50,250)}g"
    
    else:
        return f"{rd.randint(300,500)}ml"
```

## Gerando 20 registros com informações aleatórias e imprimindo no formato SQL

```python
receitas = [
    "Bolo de Açúcar e Canela",
    "Arroz com Feijão",
    "Purê de Batata",
    "Salada de Cebola e Queijo Mussarela",
    "Ovos Mexidos com Manteiga",
    "Pão de Mel com Chocolate Meio Amargo",
    "Panqueca de Farinha de Trigo e Leite Integral",
    "Cookie de Aveia e Mel",
    "Frango ao Vinagre com Pimenta-do-Reino",
    "Molho de Tomate com Sal e Alho",
    "Risoto de Arroz com Queijo Mussarela",
    "Feijoada Tradicional",
    "Biscoito de Amido de Milho",
    "Torta de Batata com Queijo Mussarela",
    "Bolo de Chocolate Meio Amargo",
    "Omelete de Ovo com Queijo Mussarela",
    "Pão de Queijo",
    "Batata Assada",
    "Caldo Verde",
    "Panetone com Açúcar e Canela"
]

# Começa a imprimir o comando SQL de inserção, 
# com os nomes das colunas da tabela 'ingredientes'

print("INSERT INTO receitas (nome, porcao, id_tabela_nutricional)\nVALUES",end="")

for i in range(20):
    nome = receitas[i]
    porcao = gerar_porcao(nome)
    id_tabela_nutricional = i + 1

    # Imprime os valores formatados para inserir no banco
    if (i < 19):
        print(f"('{nome}', '{porcao}', {id_tabela_nutricional}),")
    else:
        print(f"('{nome}', '{porcao}', {id_tabela_nutricional});")

```
